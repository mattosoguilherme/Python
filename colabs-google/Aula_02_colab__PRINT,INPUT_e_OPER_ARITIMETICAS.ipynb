## #02 - Calculadora de aumento de aluguel

Vamos construir um programa que irá calcular o aumento anual do seu aluguel em duas partes:

### Parte 1
A sua calculadora vai receber o `valor do aluguel` e calcular o aumento baseado no `IGPM de 31%`. A calculadora deve apresentar o aluguel reajustado no formato `R$ XXXX.XX`

**Exemplo:**
```
Valor do aluguel = 1000
Valor do aluguel reajustado = R$ 1310,00

````python
aluguel = float(input('digite o valor do aluguel:R$'))# input para coletar valor da variavel 
IGPM = 0.31 #taxa do percentual a ser calculado
reajuste = aluguel * IGPM #operação aritimetica(multiplicação) para calcular valor do reajuste
total = reajuste + aluguel #operacação aritimetica(soma) para calcular resultado da var reajuste + var aluguel

total = f'{total:_.2f}' 
total = total.replace('.',',').replace('_',".")
aluguel =f'{aluguel:_.2f}'
aluguel = aluguel.replace('.',',').replace('_',".")

print(f'\nValor do aluguel = R$ {aluguel}\nValor do aluguel reajustado:R$ {total}')
````

### Parte 2
Agora, altere sua calculadora para receber além do `valor do aluguel`, o percentual do reajuste no formato `XX%`.  

**Dica:** Descubra uma forma de transformar o percentual recebido em um número para efetuar o cálculo.

**Exemplo:**
```
Valor do aluguel = 1000
Percentual do reajuste = 31%
Valor do aluguel reajustado = R% 1310,00
```

````python
aluguel=float(input('digite o valor do aluguel:R$'))
IGPM=float(input('Digite o percentual do reajuste: '))
taxa=IGPM /100
reajuste=aluguel*taxa
valorFinal=reajuste+aluguel

valorFinal = f'{valorFinal:_.2f}'
valorFinal = valorFinal.replace('.',',').replace('_','.')
aluguel = f'{aluguel:_.2f}'
aluguel = aluguel.replace('.',',').replace('_','.')

print(f'\nValor do aluguel = R${aluguel}\nPercentual do reajuste = {taxa:.0%}\nValor do aluguel reajustado = R${valorFinal}')
````
